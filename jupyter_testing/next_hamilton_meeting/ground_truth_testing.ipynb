```python
import matplotlib.pyplot as plt
import model_utility_rpjb
import glob
from tensorflow import keras
import numpy as np
import segmentation_models as sm
import preprocess_filter
```

```python
fig, axes = plt.subplots(2, 3, figsize = (14, 8))

model_result_list = ["models/s3_compare/fc_no_aug_norm_spoke_15im_100e_resnet34.json", 
                 "models/s3_compare/fc_no_aug_norm_spoke_15im_1000e_resnet34.json",
                 "models/s3_compare/fc_no_aug_norm_spoke_15im_5000e_resnet34.json",
                 "models/s3_compare/c_no_aug_norm_spoke_15im_100e_resnet34.json",
                 "models/s3_compare/c_no_aug_norm_spoke_15im_1000e_resnet34.json",
                 "models/s3_compare/c_no_aug_norm_spoke_15im_5000e_resnet34.json"]
for model_result, axis in zip(model_result_list,axes.flatten()):
    split_ = model_result.split('/')[-1].split('_')
    title = split_[0] + "_" + split_[6]
    

    model_utility_rpjb.display_accuracy_in_panel(model_result, title, axis)

```

```python


model_path_list = ["models/s3_compare/fc_no_aug_norm_spoke_15im_100e_resnet34.h5", 
                 "models/s3_compare/fc_no_aug_norm_spoke_15im_1000e_resnet34.h5",
                 "models/s3_compare/fc_no_aug_norm_spoke_15im_5000e_resnet34.h5",
                 "models/s3_compare/c_no_aug_norm_spoke_15im_100e_resnet34.h5",
                 "models/s3_compare/c_no_aug_norm_spoke_15im_1000e_resnet34.h5",
                 "models/s3_compare/c_no_aug_norm_spoke_15im_5000e_resnet34.h5"]

test_rpjb_path = "/datasets/spokes_test_rpjbs/W1593676089_1_CALIB.rpjb"

plt.imshow(preprocess_filter.apply_filters(test_rpjb_path, True), cmap='gray', origin = 'lower')
plt.show()

plt.imshow(preprocess_filter.apply_filters(test_rpjb_path, False), cmap='gray', origin = 'lower')
plt.show()

fig, axes = plt.subplots(2, 3, figsize = (14, 8))
print(test_rpjb_path)

filter_imgs = False
for model_path, axis in zip(model_path_list,axes.flatten()):
    print(model_path)
    model = keras.models.load_model(model_path, compile = False)
    model.compile(optimizer = "Adam" , loss = "binary_crossentropy", metrics = [sm.metrics.IOUScore()], )

    split_ = model_path.split('/')[-1].split('_')
    title = split_[0] + "_" + split_[6]
    if split_[0] == 'fc':
        filter_imgs = True
    else:
        filter_imgs = False
    print(split_[0])


    model_utility_rpjb.display_model_result_in_panel(model, test_rpjb_path, title, axis, filter_imgs)

plt.subplots_adjust(hspace=-0.5)
```
