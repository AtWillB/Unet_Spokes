```python
test_folder = glob.glob(f"filtered/*.png")

# for path in test_folder:
#     # print(path)
pixel_values_list = []

target_size = (736, 160)

for img_path in test_folder:
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    if img is not None:
        print(img_path)
        # Resize image to target size
        resized_img = cv2.resize(img, target_size)
        plt.imshow(resized_img, cmap="gray")
        plt.show()
        pixel_values_list.append(resized_img)

        
pixel_values_array = np.array(pixel_values_list)

pixel_values_array = normalize(pixel_values_array, axis=1)
print(pixel_values_array.shape)
```

```python
model_results = []

for img in  pixel_values_list:
    img = img.reshape((1, 160, 736))
    prediction = model.predict(img)
    prediction = prediction.reshape((160, 736))
    plt.imshow(prediction, cmap = "gray")
    plt.show()

    prediction[prediction > .30] = 1
    prediction[prediction <= .30] = 0
    prediction = prediction.astype(np.uint8)

    model_results.append(prediction)

model_results = np.array(model_results)
```
